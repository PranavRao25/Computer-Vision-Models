# How to write a model architecture

## Fully Connected Network

### Linear Layers:

```python
nn.Linear(in_features, out_features)
```
: A generic linear layer with weights and biases ($y=xA^{T}+b$)
```python
nn.Bilinear(in_features1, in_features2, out_features)
```
: A generic bilinear layer with two inputs and one output ($y=x_{1}​^{T}Ax_{2}​+b$)

### Activation Functions

```python
nn.ReLU() / nn.LeakyReLU() / nn.LogSigmoid() / nn.Sigmoid() / nn.Tanh() / nn.Softmax()
```

```python
nn.Softplus(beta, threshold)
```

## Convolutional Network

### Convolution Layers:

```python
nn.Conv1d(in_channels, out_channels, kernel_size, stride, padding, dilation) / nn.Conv2d(...) / nn.Conv3d(...)
```
: A generic convolutional layer of required dimension with the relation between input and output vector dimensions being:

Input: $(N,Cin,Lin)$ or $(Cin,Lin)$<br>
Output: $(N,Cout,Lout)$ or $(Cout,Lout)$<br>
$Lout=⌊\frac{Lin+2×padding−dilation×(kernel−1)−1}{stride}+1⌋$

```python
nn.ConvTranspose1d(in_channels, out_channels, kernel_size, stride, padding, output_padding, dilation) / nn.ConvTranspose2d(...) / nn.ConvTranspose3d(...)
```
: Applies a Reverse / Transpose convolution

Input: $(N,Cin,Lin)$ or $(Cin,Lin)$<br>
Output: $(N,Cout​,Lout​)$ or $(Cout​,Lout​)$, where<br>
$Lout=(Lin−1)×stride−2×padding+dilation×(kernel−1)+output\_padding+1$<br>

### Pooling Layers

```python
nn.MaxPool1d(kernel_size, stride, padding,dilation) / nn.MaxPool2d(...) / nn.MaxPool3d(...)
```
: apply max pool operation on the vector

```python
nn.AvgPool1d(kernel_size, stride, padding, dilation) / nn.AvgPool2d(...) / nn.AvgPool3d(...)
```

```python
nn.AdaptiveMaxPool1d(output_size) / nn.AdaptiveMaxPool2d(...) / nn.AdaptiveMaxPool2d(...) / nn.AdaptiveAvgPool1d(...)
```

### Normalization Layers

```python
nn.BatchNorm1d(num_features, eps, momentum) / nn.BatchNorm2d(...) / nn.BatchNorm3d(...)
```
: batch normalization

```python
nn.GroupNorm(num_groups, num_channels, eps)
```
: group normalization over the given batch

```python
nn.LayerNorm(normalized_shape, eps)
```
: layer normalization

```python
nn.InstanceNorm1d(num_features, eps, momentum) / nn.InstanceNorm2d(...) / nn.InstanceNorm3d(...)
```
: Instance normalization

## Recurrent Network

```python
nn.RNN(input_size, hidden_size, num_layers, nonlinearity, bias, dropout, bidirectional)
nn.LSTM(input_size, hidden_size, num_layers, bias, dropout, bidirectional, proj_size) / nn.GRU(...)
```

### Transformer

```python
nn.Transformer(d_model, nhead, num_encoder_layers, num_decoder_layers, dim_feedforward, dropout, activation)
nn.TransformerEncoder(encoder_layer, num_layers, norm)
nn.TransformerDecoder(decoder_layer, num_layers, norm)
```
